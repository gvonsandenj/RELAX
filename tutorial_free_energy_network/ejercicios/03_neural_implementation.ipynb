```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
u = 2.0           # sensory input mean
sigma_u = 1.0     # sensory input variance
stdv_u = np.sqrt(sigma_u)

v_p = 3.0         # food size prediction mean
sigma_p = 1.0     # food size prediction variance
stdv_p = np.sqrt(sigma_p)
```

```python
def g(v):
    return np.pow(v,2)

def g_prime(v):
    return 2*v




phi_list = []
epsilon_p_list = []
epsilon_u_list = []

total_time = 5
delta_t = 0.01
steps = int(5/delta_t)
t = np.linspace(0,total_time,steps)


# Initial state
phi = v_p
epsilon_p = 0
epsilon_u = 0


# Update loop
for i in t:
    phi_list.append(phi)
    epsilon_p_list.append(epsilon_p)
    epsilon_u_list.append(epsilon_u)

    phi_prev = phi
    epsilon_u_prev = epsilon_u
    epsilon_p_prev = epsilon_p

    phi = phi_prev + delta_t * (epsilon_u_prev * g_prime(phi_prev) - epsilon_p_prev)
    epsilon_p = epsilon_p_prev + delta_t * (phi_prev - v_p - sigma_p * epsilon_p_prev)
    epsilon_u = epsilon_u_prev + delta_t * (u - g(phi_prev) - sigma_u * epsilon_u_prev)

plt.figure(figsize=(10, 6))
plt.plot(t, phi_list, label="phi")
plt.plot(t, epsilon_u_list, label="epsilon_u")
plt.plot(t, epsilon_p_list, label="epsilon_p")

plt.xlabel("Time")
plt.ylabel("Value")
plt.legend()
plt.grid(True)
plt.show()
    
```
