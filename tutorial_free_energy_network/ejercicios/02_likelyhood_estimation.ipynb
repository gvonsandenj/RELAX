```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
u = 2.0           # sensory input mean
sigma_u = 1.0     # sensory input variance
stdv_u = np.sqrt(sigma_u)

v_p = 3.0         # food size prediction mean
sigma_p = 1.0     # food size prediction variance
stdv_p = np.sqrt(sigma_p)

```

```python
def g(v):
    return np.pow(v,2)

def g_prime(v):
    return 2*v

def gradient(phi):
    return (v_p - phi) / sigma_p + (u - g(phi)) / sigma_u * g_prime(phi)


phi_list = []
total_time = 5
delta_t = 0.01
steps = int(5/delta_t)

t = np.linspace(0,total_time,steps)

phi = v_p # most likely value initialization

# Update loop
for i in t:
    phi_list.append(phi)
    phi = phi + gradient(phi) * delta_t

plt.plot(t, phi_list)
plt.axhline(phi, color='gray', linestyle='dotted')
plt.text(t[-1], phi, f'{phi:.3f}', va='bottom', ha='right')
plt.ylim(-2, 3.5)
plt.xlabel('time')
plt.ylabel('$\phi$')
plt.show()
```
