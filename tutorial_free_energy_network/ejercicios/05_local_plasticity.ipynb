```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
mean_phi = 5 # mean of input from the current level
Sigma_phi = 2 # variance of input from the current level
stdv_phi = np.sqrt(Sigma_phi)
phi_above = 5 # input from the level above
DT = 0.01 # integration step
MAXT = 20 # maximum time considered
TRIALS = 1000 # number of simulated trials
LRATE = 0.01 # learning rate

N_STEPS = int(MAXT / DT) # number of integration steps per trial

Sigma_list = [1] # Sigma(1) = 1

phi_list = []

# trial 1 already fixed (Sigma_list[0]), so the loop computes trials 2..TRIALS
for trial in range(1, TRIALS):
    phi = mean_phi + stdv_phi * np.random.randn()
    phi_list.append(phi)

    epsilon_list = [0] # error(1) = 0
    e_list = [0] # e(1) = 0

    for i in range(1, N_STEPS):
        error = epsilon_list[i-1] + DT * (phi - phi_above - e_list[i-1])
        e = e_list[i-1] + DT * (Sigma_list[trial-1] * epsilon_list[i-1] - e_list[i-1])

        epsilon_list.append(error)
        e_list.append(e)

    Sigma = Sigma_list[trial-1] + LRATE * (epsilon_list[-1] * e_list[-1] - 1)
    Sigma_list.append(Sigma)

plt.plot(Sigma_list, 'k')
plt.xlabel('Trial')
plt.ylabel(r'$\Sigma$')
plt.show()





```
