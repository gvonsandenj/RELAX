```python
import numpy as np
import matplotlib.pyplot as plt

from scipy.stats import Normal, norm
from scipy.integrate import quad
```

```python
u = 2.0           # sensory input mean
sigma_u = 1.0     # sensory input variance
stdv_u = np.sqrt(sigma_u)

mu_p = 3.0         # food size prediction mean
sigma_p = 1.0     # food size prediction variance
stdv_p = np.sqrt(sigma_p)


def integrand_1(v):
    p_v = Normal(mu=mu_p,sigma= stdv_p).pdf(v)
    p_u_given_v = Normal(mu=v**2, sigma = stdv_u).pdf(u)
    return p_v * p_u_given_v

def integrand_2(v,p_u):
    p_v = Normal(mu=mu_p,sigma= stdv_p).pdf(v)
    p_u_given_v = Normal(mu=v**2, sigma = stdv_u).pdf(u)
    return p_v * p_u_given_v / p_u





p_u, estimated_error = quad(integrand_1, 0.01,5.0)


v = np.arange(0.01,5,0.01)
p_v_given_u = integrand_2(v,p_u=p_u)

plt.plot(v,p_v_given_u)
```
